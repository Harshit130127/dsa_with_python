## Dynamic arrays


```python
import ctypes
```

```python
class MyList:

    def __init__(self):
        self.size = 1  # capacity of the list
        self.n = 0  # current item in the list

        # create a c type array with size = self.size

        self.A = self.__create__array(self.size)

        def __create_array(self, size):
            # create a c type array with fixed size or capacity 
            return (size * ctypes.py_object)()
```
