## Dynamic arrays


```python
import ctypes
```

```python
class MyList:

    def __init__(self):
        self.size = 1  # capacity of the list
        self.n = 0  # current item in the list

        # create a c type array with size = self.size

        self.A = self.__create_array(self.size)   # it is creating array A


    def append(self, item):
        if self.n == self.size:
            # resize the array
            self.__resize(2 * self.size)

        # append
        self.A[self.n] = item
        self.n += 1


    def pop(self):
        if self.n == 0:
            return "empty list"


        print(self.A[self.n - 1])
        self.n = self.n -1


    def clear(self):
        self.size = 1
        self.n = 0




    def __resize(self, new_size):
        # create a new array with new capacity
        B = self.__create_array(new_size)
        self.size = new_size


        # copy the content of the old array to new array
        for i in range(self.n):
            B[i] = self.A[i]

        self.A = B

    def __len__(self):
        return self.n


    def __str__(self):
        result = ''
        for i in range(self.n):
            result = result + str(self.A[i]) + ','

        return '[' + result[:-1] + ']'

    def __getitem__(self, index):
        if index < 0 or index >= self.n:
            raise IndexError('Index out of bounds')
        return self.A[index]

    def __create_array(self, size):
            # create a c type array(referential array) with fixed size or capacity
            return (size*ctypes.py_object)()
```

```python
L = MyList()
```

```python
L.append('hello')
L.append(3.4)
L.append(True)
L.append(100)
```
